```python
cannabis_keywords = ['cannabis', 'wiet', 'zaadleverancier', 'smartshop']
greenery_keywords = ['park', 'stadspark', 'botanische tuin', 'tuin']
hotel_keywords = ['hotel', 'herberg', 'jeugdhostel', 'accomodatie', 'serviceaccommodatie', 'kampeerterrein']
restaurant_keywords = ['restaurant', 'steakhouse', 'ontbijt', 'lunch', 'vis', 'hamburger', 'pizzeria', 'bar', 'koffiehuis', 'cafe', 'bakkerij', 'eetcafe', 'brasserie', 'pub', 
                       'ijssalon', 'sushi takeaway', 'koffiebranders', 'bistro', 'chocolaterie', 'patisserie', 'creperie', 'foodcourt', 'brouwerij', 'afhaalpizza', 'lounge', 
                       'banketbakker', 'kebabzaak', 'bierzaal']
store_keywords = ['winkel', 'boetiek', 'mode', 'juwelier', 'warenhuis', 'cadeau', 'parfumerie', 'e-sigaretten en accessoires', 'horlogerie', 'kleding', 'markt']
museum_keywords = ['museum', 'kunst', 'cultureel centrum', "dierentuin"]
tourist_keywords = ['escaperoomcentrum', 'amusementscentrum', 'midgetgolfbaan', 'bowlingbaan', 'poolcentrum', 'schaatsbaan', 'bioscoop']
monument_keywords = ['historisch herkenningspunt', 'monument', 'kerk', 'molen', 'brug', 'beeldhouwwerk', 'erfgoedbescherming', 'kasteel', 'gedenkteken', 'synagoge', 
                     'hindoeistische tempel', 'historische plaats', 'herdenkingsplek', 'toeristische attractie', "herdenkingspark"]
nightclub_keywords = ['nachtclub', 'club', 'discotheek']
concert_keywords = ['concertzaal', 'locatie voor live muziek', 'stadion', 'theater', 'kindertheater', 'evenementenlocatie']
conference_keywords = ['conferentiecentrum', 'congrescentrum']


# Ensure all values are strings before applying `.lower()`
df['category'] = df['main_category'].astype(str)



# Categorize cannabisshops
df['category'] = df['category'].apply(
    lambda x: 'Cannabisshop' if any(keyword in x.lower() for keyword in cannabis_keywords) else x
)

# Categorize hotels
df['category'] = df['category'].apply(
    lambda x: 'Hotel' if any(keyword in x.lower() for keyword in hotel_keywords) else x
)
                       
# Categorize restaurants
df['category'] = df['category'].apply(
    lambda x: 'Dining & Cafes' if any(keyword in x.lower() for keyword in restaurant_keywords) else x
)

# Categorize stores
df['category'] = df['category'].apply(
    lambda x: 'Shopping' if any(keyword in x.lower() for keyword in store_keywords) else x
)

# Categorize museum
df['category'] = df['category'].apply(
    lambda x: 'Cultural & Historic' if any(keyword in x.lower() for keyword in museum_keywords) else x
)

# Categorize tourist atraction
df['category'] = df['category'].apply(
    lambda x: 'Activities' if any(keyword in x.lower() for keyword in tourist_keywords) else x
)

# Categorize monuments
df['category'] = df['category'].apply(
    lambda x: 'Cultural & Historic' if any(keyword in x.lower() for keyword in monument_keywords) else x
)
                     
# Categorize nightclubs
df['category'] = df['category'].apply(
    lambda x: 'Entertainment & Nightlife' if any(keyword in x.lower() for keyword in nightclub_keywords) else x
)

# Categorize concert
df['category'] = df['category'].apply(
    lambda x: 'Event venue' if any(keyword in x.lower() for keyword in concert_keywords) else x
)

# Categorize conference
df['category'] = df['category'].apply(
    lambda x: 'Conference' if any(keyword in x.lower() for keyword in conference_keywords) else x
)

# Categorize cannabisshops
df['category'] = df['category'].apply(
    lambda x: 'Green Recreation' if any(keyword in x.lower() for keyword in greenery_keywords) else x
)

```
