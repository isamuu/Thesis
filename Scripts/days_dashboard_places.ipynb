```python

# Define a reference date for a given week
reference_monday = datetime(2024, 1, 1)  # Adjust to the first Monday of the year or any reference Monday

# Create a mapping from day names to a date in the reference week
day_mapping = {day: reference_monday + timedelta(days=i) for i, day in enumerate([
    "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"
])}

# Sample DataFrame (assuming it's already loaded)
df['datetime'] = df.apply(lambda row: day_mapping[row['Dayname']] + timedelta(hours=row['Hour']), axis=1)
```

```python
home = df.merge(times, on = "place_id", how = "left")[["Buurt", "Wijk", "place_id", "reviews 2024", "name", "pressure", "datetime", "geometry"]]
```
