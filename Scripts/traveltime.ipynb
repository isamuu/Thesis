```python
# Define multiple coordinates
coordinates_list = [
    Coordinates(lat=52.312340, lng=4.947792),  # Bijlmer
    Coordinates(lat=52.3408906, lng=4.8723349),  # Zuid
    Coordinates(lat=52.4019931, lng=4.9323589),  # Buikslotermeerplein
    Coordinates(lat=52.3596541, lng=4.80387975),  # Osdorpplein
    Coordinates(lat=52.3882035, lng=4.8316848)   # Sloterdijk-Centrum
]




# Request travel time data in WKT format
results = await sdk.time_map_wkt_async(
    coordinates=coordinates_list,
    arrival_time=datetime.now(),
    transportation=Driving()
)

# **Save the full response to a JSON file instead of printing it**
with open("isochrones_response.json", "w") as f:
    json.dump([r.wkt if hasattr(r, "wkt") else None for r in results], f, indent=4)

print("Full response saved to 'isochrones_response.json'.")

# Extract relevant data into a list
data = []
for idx, (point, result) in enumerate(zip(coordinates_list, results)):
    if isinstance(result, tuple):  # If it's a tuple, extract the first item
        result = result[0]  

    polygon_wkt = result.wkt if hasattr(result, "wkt") else None  # Extract WKT safely

    result_info = {
        "id": f"point_{idx+1}",
        "latitude": point.lat,
        "longitude": point.lng,
        "travel_time": 1800,
        "isochrone_polygon_wkt": polygon_wkt  # Storing WKT polygon
    }
    data.append(result_info)

# Convert to Pandas DataFrame
df = pd.DataFrame(data)

# Save DataFrame to CSV instead of printing
df.to_csv("C:/Users/isamu/OneDrive/THESIS/DATA/isochrones_data.csv", index=False)
print("Isochrones data saved to 'isochrones_data.csv'.")

```
